```python
# divide data into data into dependent and independent variable.
x = df.iloc[:,[2,3]].values
y = df.iloc[:,-1].values
```

```python
# Splitting dataset into training and testing. 75% of data is used fro training the model and 25% data is used
# for test to perform the model.

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state = 42, test_size = 0.25)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
# # Now, it is very important to perform feature scaling here because of Age and Estimated Salary values lies in
# different ranges
# If we do not scale the features then estimated salary feature will dominate age feature when the model finds the nearest 
# neighbour to a datapoint in dat space.

# fit - training learning from the data mean and standard deviation
# transform will apply the changes to data (x-mean)/(std)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
# Convert to dataframe
x_train = pd.DataFrame(x_train, columns = ['Age','Salary'])
```

```python
# Training our Logistic Regression Model
from sklearn.linear_model import LogisticRegression
Logis_Reg = LogisticRegression()
Logis_Reg.fit(x_train, y_train)
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
y_pred = Logis_Reg.predict(x_test)
y_pred
```

```python
# confusion matrix
from sklearn.metrics import confusion_matrix
CM = confusion_matrix(y_pred, y_test)
CM
```

```python
# Performance measure-Accuracy
from sklearn.metrics import accuracy_score
AC = accuracy_score(y_pred, y_test)
print("Accuracy :", AC)
```

```python
# classification report
from sklearn.metrics import classification_report
print(classification_report(y_pred, y_test))
```

```python
# Rebuilding the model with Three Independent Variables
```

```python
# divide data into data into dependent and independent variable.
X = df.iloc[:,[1,2,3]].values
Y = df.iloc[:,-1].values
```

```python
X
```

```python
# Since our dataset containing character variables we have to encode it using Label Encoder.
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:,0] = le.fit_transform(X[:,0])
print(X[0:7,:])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state = 0, test_size = 0.20)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.fit_transform(X_test)
```

```python
# Convert into Dataframe
X_train = pd.DataFrame(X_train, columns = ['Gender','Age','Salary'])
```

```python
# Training our Logistic Regression Model
from sklearn.linear_model import LogisticRegression
Logis_Reg = LogisticRegression(random_state = 0)
Logis_Reg.fit(X_train, Y_train)
```

```python
Y_pred = Logis_Reg.predict(X_test)
Y_pred
```

```python
Y_test
```

```python
# Confusion Matrix
from sklearn.metrics import confusion_matrix
CM = confusion_matrix(Y_pred, Y_test)
CM
```

```python
# Performance measure-Accuracy
from sklearn.metrics import accuracy_score
AC = accuracy_score(Y_pred, Y_test)
print("Accuracy :", AC)
```

```python
# classification report
from sklearn.metrics import classification_report
print(classification_report(Y_pred, Y_test))
```

```python
# Building ROC Curve
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Predicted probabilities for positive class
y_prob = Logis_Reg.predict_proba(X_test)[:,1]

# Compute ROC Curve
fpr, tpr, thresholds = roc_curve(Y_test, y_prob)

# Compute AUC
auc = roc_auc_score(Y_test,y_prob)
print("AUC Score", auc)
```

```python
# Plot ROC Curve
plt.figure(figsize=(7,5))
plt.plot(fpr, tpr, label = f"AUC = {auc:.2f}")
plt.plot([0,1],[0,1],'k--', label = "Random Guess")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```
