```python
X=df.iloc[:,6:8].values
X
```

```python
Y = df.iloc[:,8].values
Y
```

```python
# Splitting the dataset into training and testing data
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.25,random_state=10)
```

```python
# Scaling - Standard Scaling (z=(X-Xmean)/std_deviation)
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
# Apply K_Nearest-Neighbour algorithm
from sklearn.neighbors import KNeighborsClassifier
obj = KNeighborsClassifier()
```

```python
obj.fit(X_train, y_train)
```

```python
y_pred = obj.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix
m1 = confusion_matrix(y_test, y_pred)
m1
```

```python
# To find Accuracy
from sklearn.metrics import accuracy_score
Accu = accuracy_score(y_test, y_pred)*100
Accu
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plt.figure(figsize=(8,6))
plot_decision_regions(X_train,y_train,clf=obj,legend = 3)
plt.title('KNN Algorithm (Training set)')
plt.xlabel('')
plt.ylabel('')
plt.show()
```
