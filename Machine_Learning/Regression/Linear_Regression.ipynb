```python
# Remove leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Check column names
print(df.columns)

# Create boxplot
plt.figure(figsize=(5,3))
sns.boxplot(df['Total Payment'])
plt.show()
```

```python
# Extract the dependent and independent varibles.
x = df.iloc[:,:1].values
y = df.iloc[:,1].values
```

```python
x.shape
```

```python
!pip install scikit-learn
```

```python
# split the data into training and testing set
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=5)
```

```python
x_train.size
```

```python
x_test.size
```

```python
from sklearn.linear_model import LinearRegression
LR = LinearRegression()
```

```python
# train the model
LR.fit(x_train, y_train)
```

```python
y_pred = LR.predict(x_test)
y_pred
```

```python
y_test
```

```python
# Calculation of error
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, y_pred)
print("The mean error is", mse)
```

```python
rmse = np.sqrt(mse)
rmse
```

```python
r_score = r2_score(y_test, y_pred)
print("The accuracy of the model is ", r_score*100)
```

```python
LR.coef_ # this is to find m(slope)
```

```python
LR.intercept_ # this is to find c(intercept)
```

```python
# Regression line on Training data
plt.figure(figsize=(5,3))
plt.scatter(x_train,y_train, color = 'g')
plt.plot(x_train, LR.predict(x_train), color = 'r')
plt.show()
```
