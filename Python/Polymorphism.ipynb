## Polymorphism

 Polymorphism in python is a term used to refer to an object's ability to take on multiple forms.
 
 The term is derived from two disteinct terms: poly, which means numerous, and morphs, which
means forms.
 
 Polymorphism is often used in Class methods, where we can have multiple classes with the same method name.
 
 Example of polymorphism is chamaeleon(girgit) who changes his color according to things he touches.

```python
class Car:
    def move(self):
        print("This for Drive")
class Boat:
    def move(self):
        print("This for Sail")
class Plane:
    def move(self):
        print("This for fly")
```

```python
Tata = Car()
Jio = Boat()
Indigo = Plane()
```

```python
Jio.move()
Indigo.move()
Tata.move()
```

```python
class India:
    def capital(self):
        print("New Delhi is capital of India")
    def language(self):
        print("Hindi is National lnguage of India")
    def type(self):
        print("India is developing country")
class USA:
    def capital(self):
        print("Washington D.C. is capital of USA")
    def language(self):
        print("English is Primary lnguage of USA")
    def type(self):
        print("USA is developed country")    
    
```

```python
Ram = India()
Jack = USA()
```

```python
Ram.language()
Jack.language()
```

```python
class Student:
    # Constructor
    def __init__(self, name, roll_no, marks):
        self.__name = name        # Private attribute
        self.__roll_no = roll_no  # Private attribute
        self.__marks = marks      # Private attribute

    # Getter methods
    def get_name(self):
        return self.__name

    def get_roll_no(self):
        return self.__roll_no

    def get_marks(self):
        return self.__marks

    # Setter methods
    def set_name(self, n):
        self.__name = n

    def set_roll_no(self, r):
        self.__roll_no = r

    def set_marks(self, m):
        if 0 <= m <= 100:
            self.__marks = m
        else:
            print("Marks should be between 0 and 100.")

    # Display method
    def display_details(self):
        print("Student Name :", self.__name)
        print("Roll Number  :", self.__roll_no)
        print("Marks        :", self.__marks)
        print("-" * 30)


# Creating 3 student objects
student1 = Student("Priya", 101, 85)
student2 = Student("Rahul", 102, 78)
student3 = Student("Anjali", 103, 92)

# Updating marks using setter method
student1.set_marks(88)
student2.set_marks(82)
student3.set_marks(95)

# Displaying student information
student1.display_details()
student2.display_details()
student3.display_details()
```
