## Inheritance

- parent and child class

- Inheritance allows us to define a class that inherits all the methods and properties from another class.

```python
class Employee:
    name = 'ABC'
    designation = 'Manager'

class Person(Employee):
    Id = 105
    Salary = 50000
```

```python
Ram = Person() 
```

```python
Ram.Salary
```

```python
Ram.name
```

```python
# Inheritance with constructor
class Employee1:
    def __init__(self,d,s):
        self.designation = d
        self.salary = s

class Person1(Employee1):
    def get(self):
        print("The salary of Employee is ", self.salary)
```

```python
Raj = Person1('Manager', 50000)
Raj.get()
```

```python
# Multilevel Inheritance

class Rect:
    def M1(self):
        print("This is Rectangle class")

class Cir(Rect):
    def M2(self):
        print("This is a Circle class")

class Sqr(Cir):
    def M3(self):
        print("This is a Square class")
```

```python
Shape = Sqr()
Shape.M2()
```

```python
Shape.M1()
```

```python
# Multiple Inheritance

class Rect:
    def M1(self):
        print("This is Rectangle class")

class Cir(Rect,Sqr):
    def M2(self):
        print("This is a Circle class")

class Sqr:
    def M3(self):
        print("This is a Square class")
```

```python
S = Cir()
S.M1()
S.M2()
S.M3()
```
