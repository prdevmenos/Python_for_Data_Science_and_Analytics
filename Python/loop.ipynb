```python
# Wap pattern for an equilateral tringle.

        #
      # # #
    # # # # #
  # # # # # # #
# # # # # # # # #

for i in range (1, 6):
    for j in range(0,5-i):
        print(' ',end=' ')
    for k in range(0,i):
        print('#',end=' ')
    for l in range(0,i-1):
        print('#',end=' ')
    print()
```
