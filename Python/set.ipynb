```python
s1 ={'apple', 'banana','google',6, 7, 9} 
s2 = {'google','microsoft','chery', 6, 56, 90}
```

```python
# intersection
s3 = s1.intersection(s2)
s3
```

```python
# union
s4 = s1.union(s2)
s4
```

```python
# symmetric_difference
s5 = s1.symmetric_difference(s2)
s5
```

```python
# difference
s6 = s1.difference(s2)
s6
```

```python
help(s1.difference)
```

```python
#  update
S  = {1,2,3,4}
T = (5,6,7,8,9,10)
S.update(T) # current value updated to new value it does not store in new variable.
S
```

```python
# Symmetric_difference_update
s1.symmetric_difference_update(s2) # current value updated to new value it does not store in new variable.
s1
```
