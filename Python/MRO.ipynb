# (MRO) - Method Resoution Order

</center>

- Method Resoltuion Order is a order that is used to executes method based on class hierarchy especially multiple inheritance.

```python
class A:
    def f1(self):
        print("f1 works")

    def f2(self):
        print("f2 works")

    def show(self):
        print("It shows A")

class B:
    def f3(self):
        print("f3 works")

    def f4(self):
        print("f4 works")

    def show(self):
        print("It shows B")

class C(A,B):
    def f5(self):
        print("f5 works")

    # def show(self):
    #     print("It shows C")

# Creating Objects
obj1 = C()
obj1.show() # MRO - This object calling show() of A class as per sequence Inherited by C class.

B.show(obj1) # MRO - This type of calling deliberately calls show() of B class without talk about any sequence.
```
