## Object Oriented Concept

Object-oriented programming (OOP) is a method of structuring a program by bundling related properties and behaviors into
individual objects.

The main concept of object-oriented Programming (OOPs) or oops concepts in Python is to bind the data and the functions
that work together as a single unit so that no other part of the code can access this data

A class is a blueprint of objects. class = attribute/data + function/methods

A object is instance(example) of a class

```python
class Animal:
    weight = 45
    color = 'Pink'
```

```python
Dog = Animal()
```

```python
Dog.weight
```

```python
Dog.color
```

```python
Dog.color = 'Brown'
Dog.color
```

```python
Wolf = Animal()
```

```python
Wolf.weight
```

```python
Wolf.color
```

## Encapsulation

Encapsulation is about protecting data inside a class.
It means keeping data (properties) and methods together in a class, while controlling how the data can be accessed from outside the class.
This prevents accidental changes to your data and hides the internal details of how your class works.

## Constructor

```python
class Car:
    def __init__(self,m,p):
        self.model = m
        self.price = p
    def display(self):
        print('The price of car is ', self.price)

# display() used print() always.
```

```python
Tata = Car('nexon', 1148415)
Tata.display()
```

```python
class Student:
    def __init__(self,s,m,i):
        self.subject = s
        self.marks = m
        self.id = i
    def display(self):
        print('The student id is', self.id, 'and marks is', self.marks, 'on', self.subject)
```

```python
Raj = Student('Maths', 91, 101)
Raj.display()
```

## Abstraction

Abstraction is the process of hiding unnecessary details and showing only the important features of something.


Example:
ATM machine → You withdraw money without knowing how banking servers work.  

Mobile apps → You tap icons, but don’t see the code behind them.

TV remote → Press buttons without knowing internal circuits.

```python
class Employee:
    def __detail(self,salary):
        print('The Salary is ', salary)
    def get(self,salary):
        self.__detail(salary)
```

```python
Ram = Employee()
Ram.get(5473589)
```

```python
class shape():
    def __area(self, r):
        print(3.14*r*r)
    def get(self, r):
        self.__area(r)
```

```python
Circle = shape()
Circle.get(4)
```

```python
class FoodDeliveryApp:
    def __process_payment(self):
        print("Payment processed")
    def __assign_delivery(self):
        print("Delivery assigned")
    def placed_order(self,item):
        self.__process_payment()
        self.__assign_delivery()
        print(f"{item} order placed!")
```

```python
Zomato = FoodDeliveryApp()
Zomato.placed_order('Pizza')
```
