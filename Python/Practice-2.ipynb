```python
password = 1234
total_amount = 300000
# Function created for Cash-Withdrawl.
def cash_withdrawl():
    global total_amount
    amount_withdrawl = int(input('Enter Withdrawl Amount : '))
    
    # Condition setup for withdrwal is less or equal to current account balance.
    if amount_withdrawl <= total_amount:
        remaining_amount = total_amount - amount_withdrawl
        print("Amount Withdrawl : ",amount_withdrawl, 'Remaining Balance : ',remaining_amount)
    else:
        print('Insufficient Balance !!!')
        
# Function created for Deposit in Account.    
def cash_credit():
    global total_amount
    amount_credit = int(input('Enter Deposit : '))
    update_amount = total_amount + amount_credit
    print('Account Balance : ',update_amount)

# Function created for Changing ATM Pin.
def change_password():
    global password
    new_password = int(input('Enter new pin : '))

    # Condtion setup to check new password not same as old password
    if password == new_password:
        print('Invalid')
    else:
        print('Password Changed Successfully !!!')
    print('Your password is changed !!!')

def check_balance():
    print("Available Balance : ", total_amount)

# Setting condition for Pin Verification.
for attempts in range(3):
    pin = int(input('Enter pin : '))
    
    if password == pin:
        print('Transaction follows ...')
        while True:
            print('Which task you want to perform ?')
            
            # Below prompt gives the choice to customer to perform as per requirement.
            print('''Press "1" for cash_withdrawl
                     Press "2" for cash_credit
                     Press "3" for change_password
                     Press "4" for check_balance
                     Press "5" for exit ''')
            option = int(input('Enter task to perform : '))
        
            # Setting Conditions for which task should perform.
            if option == 1:
                cash_withdrawl()
            elif option == 2:
                cash_credit()
            elif option == 3:
                change_password()
            elif option == 4:
                check_balance()
            elif option == 5:
                print("Thankyou for using ATM")
                break
            else:
                print('exit')
        break        
    else:
        remaining = 2 - attempts
        if remaining > 0:
            print(f"incorrect pin !!! {remaining} attempts left.")
        else:
            print("Screen time out")
```
