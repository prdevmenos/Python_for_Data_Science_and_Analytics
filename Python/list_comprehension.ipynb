List Comprehension

```python
SQ_list = [i*i for i in range(1,11)]
SQ_list
```

```python
# print even nos of given list.
number = [3,5,7,89,34,58,93,80,72,81]
Ev = [i for i in number if i%2==0]
Ev
```

```python
words = ['apple','banana','cherry','date']
# convert the name of fruit in upper case > 5
fruit = [i.upper() for i in words if len(i)>5]
fruit
```

```python
# find square of even numbers from range 4 to 50

Sq_list = [i*i for i in range(4,51)]
Sq_list
```

```python
# print prime numbers from 1 to 80.

num = int(input('Enter number.'))
if num == 1:
    print('not prime')
else:
    for i in range(2, num+1, 1):
        if num%i == 0:
            print('not prime')
            break
        else:
            print('prime')
            break
```
