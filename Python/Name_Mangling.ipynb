# Name Mangling

</center>

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

# Creating Object
my_account = Account(50000)

try:
    print(my_account.__balance)
except AttributeError as e:
    print(e)

# Accessing through name Mangling.
print(my_account._Account__balance)
```
