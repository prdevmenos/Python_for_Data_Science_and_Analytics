## Return Statement

The return statement in Python is used inside a function to exit the function and send a specific value or result back to the caller. Once a return statement executes, Python immediately stops executing the current function, ignoring any lines of code that come after it.
Return Statement is last statement.

```python
# years to days
def Day(x):
    return x * 365
    print('complete') # after return statement no statement excutes.
```

```python
Day(3)
```

```python
# even or odd
def EO(num):
    if num % 2 == 0:
        print('Even')
    else:
        print('Odd')
```

## *args, **kwargs

Think of it as: “I don’t know how many arguments the user will pass, so I’ll collect them into a tuple. In that case use *args ” . 
*args collects any number of positional arguments into a tuple.

*args → for many unnamed values (tuple).

**kwargs → for many named values (dict).

```python
def Add(*x):
    return sum(x)
```

```python
Add(2,6,8,945,67,76)
```

```python
def Detail(**x):
    print(x)
```

```python
Detail(name = 'P',age = 25, salary = 25000)
```

## Lambda Function

Lambda functions are small, anonymous(defined without a specific name & 
often use for short term task) functions in programming. They are typically used for 
short, single-expression operations where a named function isn't necessary, making 
them useful for simplifying code or passing functions as arguments to other functions.

```python
Mul = lambda x : x*10
```

```python
Mul(20)
```

```python
# Write a program to calculate a cube of a number.

cube = lambda y : y ** 3 
```

```python
cube(3)
```

```python
cal = lambda i,j,k : (i+j)/k
cal(27,81,9)
```

```python
# conditional statements in lambda.
X = lambda a,b : a if a > b else b
X(58,71)
```

```python
Even = lambda a : 'Even' if a % 2 == 0 else 'Odd'
Even(91)
```

```python
# Filter = filter(fun,seq)

marks = [45,78,38,69,90,23,51,17]
ML = list(filter(lambda i : i > 50, marks))
ML
```

```python
# fliter the below list which divisible by 5

lst1 = [23,56,89,23,25,78,90,100,300,400]
FL = list(filter(lambda j : j%5 == 0, lst1))
FL
```

```python
# filter all even numbers from 1 to 100.

FL1 = list(filter(lambda k : k%2 == 0, range(0,101)))
FL1
```

```python
a = tuple(range(0,101))
print(list(filter(lambda x : x%2 == 0,a)))
```

## Map Function

map function returns a new sequence after applying the given function on each element

The map() function executes a specified function for each item in an iterable. 

The item is sent to the function as a parameter. iterable:  A sequence, collection or an iterator object. 

You can send as many iterables as you like,

just make sure the function has one parameter for each iterable.

```python
# map() works as to change one data form to another data form.
```

```python
lst1
```

```python
L = set(map(lambda x: x*10, lst1))
L
```

```python
# find the length of each word in str1

str = 'In python Iteration refers to the process of repeatation'
str1 = str.split()
L2 = list(map(len, str1))
L2
```

```python
Z = list(map(lambda x: x.upper(), str1))
Z
```

```python
# to print A to Z characters using chr function and map function.
# A - Z => 65 - 90

num = tuple(map(chr,range(65,91)))
num
```

```python
r = range(97,123)
num1 = set(map(chr,r))
num1
```

## Global and local variable

Python Global variables are those which are not defined inside any function and have a global scope whereas Python local variables are those which are defined inside a function and their scope is limited to that function only. In other words, we can say that local variables are accessible only inside the function in which it was initialized whereas the global variables are accessible throughout the program and inside every function.

```python
x = 'global'
```

```python
#global
def Fun():
    print(x)
```

```python
Fun()
```

```python
def Fun1():
    l = 'local'
    print(l)
```

```python
Fun1()
```

```python
x = 'global'
```

```python
def Detail():
    global x
    print(x)
    x = 'local'
    print(x)
```

```python
Detail()
```
