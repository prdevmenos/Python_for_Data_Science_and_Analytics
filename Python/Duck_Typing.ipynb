# Duck Typing

</center>

Duck Typing where
- duck quacks we call as it is duck.
- crow quacks we call as it is duck only.
- sparrow quacks we call as it is duck only.

the above scenario told about duck typing in simple way.
Let us understand in the form of program.

```python
class Laptop: # Duck
    def build(self):
        print("Laptop builds")
        
class Desktop: # crow
    def build(self):
        print("Desktop builds")

class Tablet:  # sparrow is not quacking
    def open_pdf(self):
        print("Opening PDF")
        
class Allien:  
    def code(self, machine : Laptop):
        print('Allien building')
        machine.build()

hp = Laptop()
dell = Desktop()
lenovo = Tablet()

my_alli = Allien()
# my_alli.code(hp)
# my_alli.code(dell)
my_alli.code(lenovo)
```
