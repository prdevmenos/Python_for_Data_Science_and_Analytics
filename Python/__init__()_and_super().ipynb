# __init__() and super()

- super() method is used to calling super class method or attributes, especially when overriding happens.

```python
class A:
    def __init__(self):
        print("init of A")
    def f1(self):
        print("f1 works")

class B(A):
    def __init__(self):
        super().__init__()
        print("init of B")
    def f2(self):
        self.f1()
        # super().f1()
        print("f2 works")

obj1 = B()
obj1.f2()
```
