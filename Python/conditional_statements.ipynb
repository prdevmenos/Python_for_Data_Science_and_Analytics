```python
# Even or Odd
num1 = int(input("Enter no. "))
if num1%2 == 0:
    print(num1, 'is even')
else:
    print(num1, 'is odd')
```

```python
# num is divisible by 2 and 4
num2 = int(input("Enter number."))
if num2%2 == 0:
    print("Number is divisible by 2")
    if num2%4 == 0:
        print("Number is divisible by 2 & 4 : ", num2//2)
else:
    print("Number is odd")
```

```python
# check on otp
otp = 1234
otp1 = int(input("Enter otp."))
if otp1 == otp:
    print("otp is matched")
else:
    print("2 attempts left")
    otp2 = int(input("Enter otp."))
    if(otp2 == otp):
       print("otp is matched")
    else:
        print("1 attempt left")
        otp3 = int(input("Enter otp."))
        if(otp3 == otp):
            print("otp is matched")
        else:
            print("account blocked")
```

```python
# based on multiple cindition
# y - Y - Yes - YES
# n - N - No - NO

R = input("Enter string as yes/no :")
if R == 'y' or R =='Y' or R =='Yes' or R =='YES':
    print("You said Yes")
elif R == 'n' or R =='N' or R =='No' or R =='NO':
    print('You said No')
else:
    print("invalid.")
```

```python
# Grade Assignment
# 100-90 = A
# 90-70 = B
# 70-50 = C
# <50 = fail
```

```python
mark = int(input("Enter mark "))
if mark >= 90:
    print("A")
elif mark < 90 and mark >= 70:
    print("B")
elif mark < 70 and mark >= 50:
    print("C")
else:
    print("fail")
```

```python
# Giving 20% discount to senior citizen and if he/she is member then 30%

customer_age = int(input("Enter age :"))
member_status = input('enter member status: ? yes/no')
if customer_age >= 45:
    if member_status == 'yes':
        print("eligible for 30%")
    else:
        print("not eligible")
else:
    if member_status == 'yes':
        print("eligible for 20%")
    else:
        print("not eligible")
```
