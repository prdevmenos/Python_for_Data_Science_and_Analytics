# Pandas
- Library for data wrangling , cleaning , manipulation
- Intrinsic data alignment done through labeling.
- Functions such as merge, groupby , and join.
- Handle Missing Values.
- Data Standardisation.
- Is built on top of Numpy. 
- Fast and Efficient when it comes to data wrangling.
- Tools for reading and writing data.
- Data aggregation and transformation.

## Data Structures of Pandas
- Series : One dimensional
- Dataframe : Two dimensional

### Series
- One dimensional
- Data and labels. Labels are also called indexes.
- Data alignment is intrinsic and takes place automatically.
- Change the data alignment by indexing.

```python
import pandas as pd
```

```python
lst = [3,5,7,9,34,71,878]
```

```python
ser1 = pd.Series(lst)
ser1
```

```python
ser2 = pd.Series(lst, index = ['a','b','c','d','e','f','g'])
ser2
```

```python
# convert tuple into series
```

```python
# convert dictionary into series
dict1 = {'a' : 10, 'b': 20, 'c' : 30}
```

```python
ser3 = pd.Series(dict1)
ser3
```

```python
# Accessing data into labels
student_marks = pd.Series([99,98,97,96,95], index = ['Maths','Chemistry','English','Physics','Biology'])
student_marks
```

```python
print(student_marks['Chemistry'])
```

```python
print(student_marks['Chemistry':'Biology'])
```

```python
student_marks['Biology'] = 98
student_marks
```

```python
student_marks[1]
```

```python
dir(student_marks)
```

```python
student_marks
```

```python
# iloc =  we have to send index address
print(student_marks.iloc[2])
```

```python
# loc =  we have to send label address
print(student_marks.loc['English'])
```

```python
#### Vectorised operations
seriesA = pd.Series([1,2,3,4,5], index = ['a','b','c','d','f'])
seriesB = pd.Series([6,7,8,9,10], index = ['a','b','c','d','f'])
```

```python
seriesC = seriesA + seriesB
seriesC
```

```python
# Vetorized Operations
seriesD = pd.Series([1,2,3,4,5], index = ['a','b','c','d','f'])
seriesE = pd.Series([6,7,8,9], index = ['f','a','c','d'])
```

```python
seriesF = seriesD + seriesE
seriesF
```

```python
# handling null value
seriesF.dropna()
```

```python
seriesF.fillna(2)
```

```python
# mean, median, mode
print(seriesF.mean())
```

```python
print(seriesF.median())
```

```python
print(seriesF.mode())
```

```python
seriesF.fillna(seriesF.mean())
```

### groupby

```python
df_name = pd.DataFrame({'first':['George','Bill','Ronald','Jimmy','George'],
                        'second':['Bush','Clinton','Reagan','Carter','Washington'],
                       'score':[3,1,2,1,1]})
df_name
```

```python
g = df_name.groupby('first')
g.size() # to know the size of groupby columns how many times comes.
```

```python
g.get_group('George') # get_group is used to find group of particular data in dataframe.
```
