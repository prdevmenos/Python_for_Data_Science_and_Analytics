# Analysis using data visualization

- Univariate Analysis- Histogram, Line Chart, Boxplot, displot

- Bivariate Analysis - Scatter plot, line chart, bar chart, regression plot, join plot

- Multivariate Analysis- Heatmap, pie chart, area chart, tree map, pair plot

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
random  = np.random.randint(1,101,20)
random
```

```python
plt.figure(figsize = (5,3)) # (length,width) of line chart
plt.plot(random, color = 'r', linestyle = ':', linewidth = 1, marker = 'o', markersize = 2)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
web_customers = [123,345,670,678,890]
plt.figure(figsize = (5,3)) # (length,width) of line chart
plt.plot(web_customers, color = 'r', linestyle = ':', linewidth = 1, marker = 'o', markersize = 2)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
web_customers = [123,345,670,678,890, 567]
time_hrs = [7,8,9,10,11,12]
```

```python
plt.figure(figsize = (6,3))
plt.plot(time_hrs, web_customers, color = 'm', linestyle = ':', linewidth = 1, marker = 'o', markersize = 2)
plt.xlabel('Time')
plt.ylabel('No of Customers')
plt.title('Web Traffic')
plt.show()
```

## Multivariate Analysis

```python
web_Monday = [123,345,670,678,890,567]
web_Tuesday = [134,234,765,987,432,789]
web_Wednesday = [345,234,564,987,908,900]
time_hrs = [7,8,9,10,11,12]
```

```python
plt.figure(figsize = (7,4))
plt.plot(time_hrs, web_Monday, color = 'red', linestyle = '-', linewidth = 1, label = 'Mon')
plt.plot(time_hrs, web_Tuesday, color = 'green', linestyle = '-', linewidth = 1, label = 'Tue')
plt.plot(time_hrs, web_Wednesday, color = 'blue', linestyle = '-', linewidth = 1, label = 'Wed')
plt.xlabel('Time')
plt.ylabel('No of Customers')
plt.title('Web Traffic')
plt.legend() # it only work when label initiated in plot().
plt.show()
```

```python
# Scatter plot
```

```python
web_customers = [123,345,670,678,890, 567]
time_hrs = [7,8,9,10,11,12]
```

```python
plt.figure(figsize=(5,3))
plt.scatter(time_hrs, web_customers, color = 'r', s= web_customers)
plt.xlabel('Time')
plt.ylabel('No of Customers')
plt.show()
```

```python
web_Monday = [123,345,670,678,890,567]
web_Tuesday = [134,234,765,987,432,789]
web_Wednesday = [345,234,564,987,908,900]
time_hrs = [7,8,9,10,11,12]
```

```python
plt.figure(figsize = (7,4))
plt.scatter(time_hrs, web_Monday, color='green',alpha = 0.7, s=web_customers,label='Mon') # alpha attribute is used to decide color intensity.
plt.scatter(time_hrs, web_Tuesday, color='b',alpha = 0.5,s=web_customers, label = "Tue")
plt.scatter(time_hrs, web_Wednesday, color='m', alpha = 0.9,s=web_customers, label = "Web")
plt.xlabel('Time')
plt.ylabel('No of Customers')
plt.title('Web Traffic')
plt.style.use('grayscale')
plt.legend()
plt.show()
```

```python
# Subplot :  used to show multiple charts at a time.
```

```python
X = np.array(['A','B','C','D','E'])
Y = np.array([3,8,9,45,36])
```

```python
plt.figure(figsize = (9,5))
plt.subplots_adjust(wspace = 0.3, hspace = 0.5)

plt.subplot(2,2,1)
plt.plot(X,Y,color = 'r', marker = 'o')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Line Chart')

plt.subplot(2,2,2)
plt.scatter(X,Y, color ='m')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Scatter Chart')

plt.subplot(2,2,3)
plt.pie(Y, labels = X)
plt.title('Pie Chart')

plt.subplot(2,2,4)
plt.bar(X,Y, color = 'green')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Bar Chart')
```
