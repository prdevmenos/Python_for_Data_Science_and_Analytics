## List and Tuples

```python
# multiple variables referencing to the same object is known as aliasing.

A = ['Dinku', 45, 2.5, True]
B = A
print(A)
print(B)

A[0] = "Banana" # A and B are referencing same list therefore chanages also occured in both A and B
print(A)
print(B)
```

```python
# Cloning of list

C = A[:]
D = B[:]
print(C) # clone or copy of list A
print(D) # clone or copy of list B
```

```python
A[1] = 34
print(A)
print(C) # After cloning if we are doing changes in original list then it does not occured in original list.
```

## Set Logical Operations

```python
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
```

```python
# Find the intersections

intersection = album_set1 & album_set2
intersection
```

```python
# Find the difference in set1 but not set2

album_set1.difference(album_set2)  
```

```python
# Find the difference in set2 but not set1

album_set2.difference(album_set1)  
```

```python
# Use intersection method to find the intersection of album_list1 and album_list2

album_set1.intersection(album_set2) 
```

```python
# Find the union of two sets

album_set1.union(album_set2)
```

```python
# Check if superset

set(album_set1).issuperset(album_set2)   
```

```python
# Check if subset

set(album_set2).issubset(album_set1)    
```

```python
# Check if subset

set({"Back in Black", "AC/DC"}).issubset(album_set1) 
```

```python
# Check if superset

album_set1.issuperset({"Back in Black", "AC/DC"})   
```

## Dictionary
### What are Dictionaries?
A dictionary consists of keys and values. It is helpful to compare a dictionary to a list. Instead of being indexed numerically like a list, dictionaries have keys. These keys are the keys that are used to access values within a dictionary.

The best example of a dictionary can be accessing person's detais using the social security number.
Here the social security number which is a unique number will be the key and the details of the people will be the values associated with it.


Dictionary is a key-value pair type of collection.
- keys are immutable and unique.
- values are immutable, mutable and duplicates.
- Each key-value pair is separated by comma(,).

```python
# Create the dictionary

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
Dict
```

```python
# Access to the value by the key

Dict["key1"]
```

```python
# Access to the value by the key

Dict[(0, 1)]
```

```python
# Create a sample dictionary

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

```python
# Get value by keys

release_year_dict['Thriller'] 
```

```python
# Get value by key

release_year_dict['The Bodyguard'] 
```

```python
# Get all the keys in dictionary

release_year_dict.keys() 
```

```python
# Get all the values in dictionary

release_year_dict.values()
```

```python
# Append value with key into dictionary

release_year_dict['Graduation'] = '2007'
release_year_dict
```

```python
# Delete entries by key

del release_year_dict['Thriller']
del release_year_dict['Graduation']
release_year_dict
```

```python
# Verify the key is in the dictionary

"The Bodyguard" in release_year_dict
```
