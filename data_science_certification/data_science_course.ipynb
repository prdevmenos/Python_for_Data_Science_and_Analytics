## Note: Use else with for

If we want certain processing at the end of the loop we can use else with for loop i.e., afer completion of for loop we can add else also for doing
certain processing.

```python
for i in range(0,10,2):
    print(i)
else:
    print("Outside for loop",str(i))
```

```python
school
```

```python
# Accesing an element of the dictionary
print(school['Kumar'])
```

```python
# updating value to a key in dictonary
school['Priyank'] = 82
school
```

```python
# adding key-value pair in dictionary

school['Khushi'] = 94
school
```

```python
# deleting key-value pair in dictionary

del school['Priyank']
school
```

```python
# len() : to get count of dictionary items

len(school)
```

```python
print(len(school))
```
