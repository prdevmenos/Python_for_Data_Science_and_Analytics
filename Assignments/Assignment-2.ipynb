```python
#1
class Student:
    def __init__(self, n, r, m):
        self.__name = n
        self.__roll_no = r 
        self.__marks = m

    # Getter Methods
    def get_name(self):
        return self.__name
    
    def get_roll_no(self):
        return self.__roll_no

    def get_marks(self):
        return self.__marks

    # Setter Methods
    def set_name(self, n):
        self.__name = n

    def set_roll_no(self, r):
        self.__roll_no = r

    def set_marks(self, m):
        self.__marks = m
    
    def display_details(self):
        print(f"Name is {self.__name}")
        print(f"Roll No is {self.__roll_no}")
        print(f"Marks is {self.__marks}")    
```

```python
s1 = Student("Prathmesh", 101, 87)
s2 = Student("Prajwal", 102, 91)
s3 = Student("Prashant", 103, 97)
```

```python
s1.display_details()
s2.display_details()
s3.display_details()
```

```python
#2
class BankAccount:
    def __init__(self, b):
        self.__balance = b

    def deposit(self, amount):
        if(amount > 0):
            self.__balance += amount
            print("Deposited :", amount)
        else:
            print("Deposited Balance must be positve")

    def withdraw(self, amount):
        if(amount <= self.__balance):
            self.__balance -= amount
            print("Withdrawn : ", amount)
        else:
            print("Insufficient balance")
    def get_balance(self):
        return self.__balance
```

```python
account = BankAccount(0)

# Deposit 
account.deposit(5000)
print("Current Balance :", account.get_balance())

# Withdrawl
account.withdraw(1000)
print("Current Balance :", account.get_balance())
```

```python
#3
class Employee:
    def __init__(self, n, emp, s):
        self.name = n
        self.employee_id = emp
        self.salary = s
    def display_info(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Salary:", self.salary)
        
class Manager(Employee):
    def __init__(self, n, emp, s, d):
        super().__init__(n, emp, s)
        self.department = d

    def display_info(self):
        super().display_info()
        print("Department:", self.department)

m1 = Manager("Raj", 101, 50000, 'IT')
m1.display_info()
```

```python
#4
class Vehicle:
    def __init__(self, b, m):
        self.brand = b
        self.model = m

    def start(self):
        print("Vehicle is starting ... ")

class Car(Vehicle):
    def start(self):
        print(self.brand, self.model, "Car is starting with a key")

class Bike(Vehicle):
    def start(self):
        print(self.brand, self.model, "Bike is starting with self-satrt button")

# Create Objects
c1 = Car("Hyundai","Creta")
b1 = Bike("Royal Enfield", "Bullet")

# Calling start method
b1.start()
c1.start()
```
