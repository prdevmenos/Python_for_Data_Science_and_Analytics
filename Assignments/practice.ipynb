```python
def string_length(str1):
    count = 0
    for char in str1:
        count += 1 # count = count + 1
    return count
print(string_length('Prasanna Rahangdale'))
```

```python
def char_frequency(str2):
    dict = {}
    for i in str2:
        keys = dict.keys()
        if i in keys:
            dict[i] += 1
        else:
            dict[i] = 1
    return dict
print(char_frequency("www.google.com"))
```

```python
def char_frequency(str2):
    list1 = {}
    for i in str2:
        list1[i] = list1.get(i, 0) + 1
    result = []
    for key, value in list1.items():
        result.append((key, value))
    return result
print(char_frequency("www.google.com"))
```

```python
def char_frequency(str2):
    freq = {}

    for char in str2:
        freq[char] = freq.get(char, 0) + 1

    result = []
    for key, value in freq.items():
        result.append((key, value))

    return result
print(char_frequency("www.google.com"))
```

```python
from tkinter import *
from tkinter import messagebox

# Initial Data
password = 1234
total_amount = 300000

# Login Function
def verify_pin():
    entered_pin = pin_entry.get()

    if entered_pin.isdigit() and int(entered_pin) == password:
        login_frame.pack_forget()
        atm_frame.pack()
    else:
        messagebox.showerror("Error", "Invalid PIN")

# Withdraw Function
def cash_withdrawal():
    global total_amount

    amount = amount_entry.get()

    if amount.isdigit():
        amount = int(amount)

        if amount <= total_amount:
            total_amount -= amount
            balance_label.config(
                text=f"Current Balance: ₹{total_amount}"
            )
            messagebox.showinfo(
                "Success",
                f"₹{amount} Withdrawn Successfully"
            )
        else:
            messagebox.showerror(
                "Error",
                "Insufficient Balance"
            )

# Deposit Function
def cash_credit():
    global total_amount

    amount = amount_entry.get()

    if amount.isdigit():
        amount = int(amount)

        total_amount += amount

        balance_label.config(
            text=f"Current Balance: ₹{total_amount}"
        )

        messagebox.showinfo(
            "Success",
            f"₹{amount} Deposited Successfully"
        )

# Change PIN Function
def change_password():
    global password

    new_pin = pin_change_entry.get()

    if new_pin.isdigit():
        new_pin = int(new_pin)

        if new_pin == password:
            messagebox.showerror(
                "Error",
                "New PIN cannot be same as old PIN"
            )
        else:
            password = new_pin
            messagebox.showinfo(
                "Success",
                "PIN Changed Successfully"
            )

# Main Window
root = Tk()
root.title("ATM Management System")
root.geometry("500x450")

# ---------------- Login Frame ----------------
login_frame = Frame(root)

Label(
    login_frame,
    text="ATM Login",
    font=("Arial", 18, "bold")
).pack(pady=20)

Label(
    login_frame,
    text="Enter PIN"
).pack()

pin_entry = Entry(login_frame, show="*")
pin_entry.pack(pady=10)

Button(
    login_frame,
    text="Login",
    command=verify_pin
).pack(pady=10)

login_frame.pack()

# ---------------- ATM Frame ----------------
atm_frame = Frame(root)

Label(
    atm_frame,
    text="ATM Dashboard",
    font=("Arial", 18, "bold")
).pack(pady=10)

balance_label = Label(
    atm_frame,
    text=f"Current Balance: ₹{total_amount}",
    font=("Arial", 12)
)

balance_label.pack(pady=10)

Label(
    atm_frame,
    text="Enter Amount"
).pack()

amount_entry = Entry(atm_frame)
amount_entry.pack(pady=5)

Button(
    atm_frame,
    text="Withdraw",
    width=15,
    command=cash_withdrawal
).pack(pady=5)

Button(
    atm_frame,
    text="Deposit",
    width=15,
    command=cash_credit
).pack(pady=5)

Label(
    atm_frame,
    text="Change PIN"
).pack(pady=10)

pin_change_entry = Entry(atm_frame, show="*")
pin_change_entry.pack()

Button(
    atm_frame,
    text="Change PIN",
    width=15,
    command=change_password
).pack(pady=10)

Button(
    atm_frame,
    text="Exit",
    width=15,
    command=root.destroy
).pack(pady=10)

root.mainloop()
```

```python
def remove_nth_character(str3, n):
    first_part = str3[:n]
    last_part = str3[n+1:]
    return first_part+last_part

print(remove_nth_character('Python', 3))
print(remove_nth_character('Python', 9))
print(remove_nth_character('Python', 2))
print(remove_nth_character('Python', 0))
```

```python
def remove_nth_character1(str3, n):
    if n < 0 or n >= len(str3):
        return 'invalid string'
    return str3[:n] + str3[n+1:] 
print(remove_nth_character1("Python", 4))
print(remove_nth_character1("Python", 7))
print(remove_nth_character1("Python", -1))
```
