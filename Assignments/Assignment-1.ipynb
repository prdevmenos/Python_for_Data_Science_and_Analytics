4. How is a string different from a number in Python ?

The fundamental difference between a string (str) and a number (like int or float) comes as what the data represents and what user can do with it.
While a number is meant for mathematical calculations, a string is meant for text manipulation. Python sees them as completely different worlds.

    a. Numbers are written directly without any special characters. They represent quantitative values(quantity).
    b. Strings are sequences of characters wrapped in quotes (either single ' ' or double " "). Even if a string contains digits, the quotes tell Python to treat it as text.
