```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        self.nums = nums
        self.target = target
        i = 0
        j = len(nums) - 1
        while(i < j):
            sum = nums[i] + nums[j]
            if(sum == target):
                return(i,j)
            if(sum < target):
                i += 1
            else:
                j -= 1
```

```python
Raj = Solution()

nums = list(map(int, input().split()))
target = int(input("Enter target: "))
result = Raj.twoSum(nums, target)
print(result)
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        seen = {}

        for i in range(len(nums)):
            complement = target - nums[i]

            if complement in seen:
                return [seen[complement], i]

            seen[nums[i]] = i
```

```python
Raj = Solution()

nums = list(map(int, input().split()))
target = int(input("Enter target: "))
result = Raj.twoSum(nums, target)
print(result)
```
