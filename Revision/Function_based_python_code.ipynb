```python
# Check even or odd
```

```python
class Solution:
    def evenOrOdd(self,n):
        self.n = n
        if n%2 == 0:
            return "even"
        else:
            return "odd"

obj1 = Solution()
x = int(input("Enter a number :"))
result = obj1.evenOrOdd(x)
print(result)
```

```python
# Sum of n consecutive numbers.
```

```python
class Solution:
    def consecutiveNumber(self,n):
        self.n = n
        for i in range(n):
            res = n*(n+1)/2
        return int(res)

obj2 = Solution()
y = int(input("Enter a Number :"))
result = obj2.consecutiveNumber(y)
print(result)
```

```python
# Write a program to print n even numbers.
```

```python
class Solution:
    def evenNumber(self,n):
        self.n = n
        for i in range(0, n+1):
            if i%2 == 0:
                print(i,end = ", ")
obj3 = Solution()
z = int(input("Enter a Number :"))
result = obj3.evenNumber(z)
print(result)
```
