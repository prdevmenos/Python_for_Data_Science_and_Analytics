```python
class Student:
    def __init__(self,gname,groll):
        self.name = gname
        self.roll = groll

    def dispdetails(self):
        print(f"Student name is {self.name} and roll no is {self.roll}")

s1 = Student("Amit", 101)
s2 = Student("Namit", 103)
s3 = Student("Khushi", 104)

s1.dispdetails()
s2.dispdetails()
s3.dispdetails()
```

```python
class School:
    def __init__(self,Id,name,Class,section):
        self.id = Id
        self.name = name
        self.class1 = Class
        self.section = section

    def display_details(self):
        print("Id of student is :",self.id)
        print("Name of student is :",self.name)
        print("Class that studied :",self.class1)
        print("and Aprropriate section is :",self.section)

Student1 = School(1001,"Raj",8,'A')

Student1.display_details()
```
