# Object Oriented Programming

```python
class Student:
    name = "Guest User"
    grade = "A"

    def display_info(self):
        return f"student : {self.name}, grade : {self.grade}"

# Create Object
student1 = Student()
print(student1.display_info())

# Create another object
student2 = Student()
student2.name = "Amit"
student2.grade = "A+"
print(student2.display_info())
```

## Encapsulation

```python
class smartThermostat:
    def __init__(self, owner, initial_temp):
        self.owner = owner # Public attribute can be accessed directly.
        self.__temperature = initial_temp # private attribute can not excess externally.

    # 1. GETTER: Safely read the private data
    def get_temperature(self):
        return f"Current temperature is {self.__temperature}°C"

    # 2. SETTER: Safely modify the private data with validation rules
    def set_temperature(self, new_temp):
        # Validation logic for protecting the data
        if 15 <= new_temp <= 30:
            self.__temperature = new_temp
            return(f"Temperature successfully updated to {new_temp}°C.")
        else:
            return("Error: Safe temperature limits must be between 15°C and 30°C!")

my_temp = smartThermostat("John", 22)
print(my_temp.get_temperature())
print(my_temp.set_temperature(25))
print(my_temp.set_temperature(29))
```

## Abstraction

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started. Vroom!"

# my_vehicle = Vehicle() Can not able to create instance of abstract class

my_car = Car()
print(my_car.start_engine())
```

## Inheritance

```python
class Animal:
    def eat(self):
        return "Eating Food"

class Dog(Animal):
    def bark(self):
        return "Woof!"

my_dog = Dog()
print(my_dog.eat())
print(my_dog.bark())
```

### Single(Basic) Inheritance

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def work(self):
        return f"{self.name} is working."

class Developer(Employee):
    def debug_code(self):
        return f"{self.name} is fixing bugs"

dev = Developer("Ram", 95000)
print(dev.work())
print(dev.debug_code())
print(dev.salary)
```

### Multiple Inheritance

```python
class Engine:
    def start_engine(self):
        return "Engine Started"

class Electronics:
    def turn_on_lights(self):
        return "Lights On"

# Creating Inherited Class
class Car(Engine, Electronics):
    pass

my_car = Car()
print(my_car.start_engine())
print(my_car.turn_on_lights())
```

### Multilevel Inheritance

```python
class Grandparent:
    def gold(self):
        return "Family Gold"

class Parent(Grandparent):
    def house(self):
        return "Family House"

class Child(Parent):
    def car(self):
        return "My Car"

me = Child()
print(me.gold())
print(me.house())
print(me.car())
```

### Hierarchial Inheritance

```python
class Vehicle:
    def info(self):
        return "This is a Vehicle."
        
class Truck(Vehicle):
    pass

class Motorcycle(Vehicle):
    pass

# Create a instance based on child class.
semi = Truck()
bike = Motorcycle()

print(semi.info())
print(bike.info())
```

### Hybrid Inheritance

```python
class A:
    def action(self):
        return "Action from A"

class B(A):
    pass
class C(A):
    pass

class D(B,C):
    pass

# Creating Instance of a class.
obj = D()
print(obj.action())
```

## Polymorphism

```python
class Cat:
    def make_sound(self):
        return "Meow"

class Dog:
    def make_sound(self):
        return "Bark"

# A single interface/function handling different object types
def play_sound(animal_object):
    print(animal_object.make_sound())

play_sound(Cat())  
play_sound(Dog())
```

```python
class School:
    def __init__(self, student, teacher, staff):
        self.__student = student
        self.__teacher = teacher
        self.__staff = staff

    def get_student(self):
        return self.__student

    def get_teacher(self):
        return self.__teacher

    def get_staff(self):
        return self.__staff

    def set_student(self, stu):
        self.__student = stu

    def set_teacher(self, t):
        self.__teacher = t

    def set_staff(self, sta):
        self.__staff = sta

    def display_info(self):
        print("student_name :",self.__student)
        print("teacher_name :", self.__teacher)
        print("staff_name :", self.__staff)
        print("-" * 30)

# Creating an object
info1 = School("Raj", "Purshottam", "Manikram")
info2 = School("Rajesh", "Priya", "Dhaniram")
info3 = School("Anirudh", "Pushpanjali", "Ramprakash")

# Updating Student name using setter methods
info1.set_student("John")
info2.set_student("Alice")
info3.set_student("Martin")

info1.display_info()
info2.display_info()
info3.display_info()
```

```python
class India:
    def language(self):
        print("Hindi is the national language of India")
    def capital(self):
        print("'New Delhi' is the capital of India")

class USA:
    def language(self):
        print("English is the national language of USA")
    def capital(self):
        print("'Washington DC' is the capital of USA")

class Japan:
    def language(self):
        print("Japanese is the national language of Japan")
    def capital(self):
        print("'Tokyo' is the capital of Japan")

# Creating Objects of above classes
Ram = India()
John = USA()
Chang = Japan()

Ram.language()
John.language()
Chang.language()
```

```python
# Inheritance

class Rectangle:
    def S1(self):
        return("The shape is Rectangle")

class Triangle(Rectangle):
    def S2(self):
        return("The shape is Triangle")

class Circle(Triangle):
    def S3(self):
        return("The shape is Circle")

shape = Circle()
print(shape.S1())
print(shape.S2())
print(shape.S3())
```

```python
# Single Inheritance with Constructor
class School:
    def __init__(self, student, class_name, quantity):
        self.student = student
        self.class_name = class_name
        self.quantity = quantity

class Details(School):
    def display_details(self):
        print("Monitor :", self.student)
        print("Class :", self.class_name)
        print("Strength :", self.quantity)
        print("-" * 15)

d1 = Details("Ram", "VIII", 75)
d2 = Details("Raj", "IX", 70)
d3 = Details("Pooja", "XI", 72)
d1.display_details()
d2.display_details()
d3.display_details()
```

```python
class Shape:
    def __init__(self, area, perimeter):
        self.area = area
        self.perimeter = perimeter

class Circle(Shape):
    def measure1(self):
        print("Area of circle is pir2",self.area)
        print("Perimeter of circle is 2pir", self.perimeter)

class semiCircle(Shape):
    def measure2(self):
        print("Area of Semicircle is pir2/2", self.area)
        print("Perimeter of Semicircle is pir", self.perimeter)

c1 = Circle(24,245)
s1 = semiCircle(12, 123)

c1.measure1()
s1.measure2()
```
