## Datetime Library

```python
import datetime
```

```python
now = datetime.datetime.now()
print("Current date and time :", now)
```

```python
from datetime import date

current_date = date.today()
print("Today's date according to Gregorian calendar is : ", current_date)
```

```python
current = date.today()
print(current)
print(15*"-")
print("Current year is :", current.year)
print("Current month is :", current.month)
print("Current day is :", current.day)
```

```python
today = datetime.datetime.now()
print("Current Date and Time :",today)
```

```python
from datetime import time

current_time = time(20,4,57)
print("Current Hour is",current_time.hour)
print("Current Minute is",current_time.minute)
print("Current Second is",current_time.second)
print("Current MicroSecond is", current_time.microsecond)
```

```python
from datetime import datetime

display = datetime(2026, 7, 28, 20, 11, 28)

print("year =", display.year)
print("month =", display.month)
print("day =", display.day)
print("hour =", display.hour)
print("minute =", display.minute)
print("second =", display.second)
print("timestamp =", display.timestamp)
```

```python
from datetime import timedelta

one_time = timedelta(weeks = 3, days = 7, hours = 12, seconds = 43)
second_time = timedelta(days = 2, hours = 18, minutes = 8, seconds = 54)

difference = one_time - second_time

print("difference of days is :", difference)
```

```python
from datetime import date

current_date = date.today()
convert_to_string = date.isoformat(current_date)

print("String Representation of current date is :", convert_to_string)
print(type(convert_to_string))
```
