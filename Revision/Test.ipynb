```python
# Reverse a String
s1 = "Raj"
r = s1[::-1]
print(r)
```

```python
# Check Palindrome
s2 = "madam"
result = s2[::-1]
if s2 == result:
    print("Palindrome ... ")
else:
    print("Not Palindrome !")
```

```python
# Check even and odd number.
n = int(input("Enter a number"))
if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Print Fibonacci Numbers

n = int(input("Enter a Number :"))
a = 0
b = 1
for i in range(n):
    print(a, end = " ")
    c = a + b
    a = b
    b = c
```

```python
# Print Largest of two Numbers.
a1 = int(input("Enter number :"))
b1 = int(input("Enter number :"))

if a1 > b1:
    print(a1)
else:
    print(b1)
```

```python
# Check Positive, Negative and Zero

n1 = int(input("Enter a Number :"))
if (n1 > 0):
    print("It is positive")
elif (n1 < 0):
    print("It is negative")
else:
    print("It is zero")
```
