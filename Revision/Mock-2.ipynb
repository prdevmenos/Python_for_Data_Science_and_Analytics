```python
def second_largest(arr):
    if len(arr) < 2:
        return None

    lar = float('-inf')
    second_lar = float('-inf')

    for num in arr:
        if num > lar:
            second_lar = lar
            lar = num
        elif num > second_lar and num != lar:
            second_lar = num
    return second_lar if second_lar != float('-inf') else None
```

```python
second_largest((4,7,78,90))
```
