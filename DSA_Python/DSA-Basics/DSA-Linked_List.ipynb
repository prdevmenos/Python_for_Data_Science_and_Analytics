# Linked List

## Creating a Node

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

node = Node(10)

print(node.data)
print(node.next)
# print(node.data)
```

## Create and Traversed a Linked List

```python
class Node1:
    def __init__(self, data):
        self.data = data
        self.next = None

first = Node1(10)
second = Node1(20)
third = Node1(30)
fourth = Node1(40)

first.next = second
second.next = third
third.next = fourth

current = first

while current:
    print(current.data)
    current = current.next
```

## Insert Node at beginning 

```python
class Node2:
    def __init__(self,data):
        self.data = data
        self.next = None

head = Node2(20)
head.next = Node2(30)
# Create variable new_node
new_node = Node(10)
new_node.next = head
head = new_node

current = head

while current:
    print(current.data)
    current = current.next
```

## Insert Node at End

```python
class Node3:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node3(5)
tail = Node3(10)

head.next = tail
# Create variable new_node
new_node = Node3(15)
tail.next = new_node

current = head

while current:
    print(current.data)
    current = current.next

# The above code is executed well but not perfect.
```

## Insertion at end using tail pointer

```python
# Efficient Solution with Time Complexity O(1).

class Node3:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node3(5)
tail = head

new_node = Node3(10)
tail.next = new_node
tail = new_node

new_node = Node3(15)
tail.next = new_node
tail = new_node

new_node = Node3(20)
tail.next = new_node
tail = new_node

current = head

while current:
    print(current.data)
    current = current.next
```

## Insertion at end using two classes Node & LinkedList.

```python
class Node:
    def __init__(self, data):
        # To store Nodes
        self.data = data
        # To store reference of next node 
        self.next = None

class LinkedList:
    def __init__(self):
        # this is empty head node.
        self.head = None

    def Insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        
        current = self.head

        while current.next:
            current = current.next

        current.next = new_node

    def display(self):
        current = self.head

        while current:
            print(current.data, end =" -> ")
            current = current.next
        print("None")  
```

```python
ll = LinkedList()

ll.Insert_at_end(10)
ll.Insert_at_end(20)
ll.Insert_at_end(30)
ll.Insert_at_end(40)
ll.Insert_at_end(50)
ll.Insert_at_end(60)

ll.display()
```
