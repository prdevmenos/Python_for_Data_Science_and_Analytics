# Arrays In Python

```python
from array import array
```

```python
numbers = array('i',[10,20,30,40,50])
print(numbers)
```

## Traversing Elements

```python
# Find Largest Number

numbers = [10,50,20,80,40]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
print(largest)
```

```python
# Find Smallest Number

numbers1 = [10,50,20,80,40]

for num1 in numbers1:
    if num1 < numbers[0]:
        num1 = numbers[0]

print(numbers[0])
```

```python
# Sum of all Elements

number2 = [10,50,20,80,40]
sum1 = 0

for num2 in number2:
    sum1 = sum1 + num2
    print(sum1)

print("Sum of all Elements :",sum1)
```

## Reversing Elements 

```python
# Reversing Array

arr = [10,20,30,40,50]
arr2 = []

for i in range(len(arr)-1, -1, -1):
    print(arr[i])
    arr2.append(arr[i])

print(arr2)
```

```python
# Alternate Solution

arr = [10,20,30,40,50]
print(arr[::-1])
```

## Removing Elements in Array of Specific Position

```python
# Removing an element without remove()

arr1 = [89, 56, 90, 31, 15, 78]
arr3 = []

for i in range(len(arr1)):
    if i != 3:
        arr3.append(arr1[i])

print(arr3)
```

```python
arr4 = []
for item in arr1:
    if item != 31:
        arr4.append(item)

print(arr4)
```

```python
arr1 = [89, 56, 90, 31, 15, 78]
arr6 = []

for i in range(len(arr1)):
    if i == 3:
        continue
    arr6.append(arr1[i])

print(arr6)
```

```python
arr5 = []
for item in arr1:
    if item == 15:
        continue
    arr5.append(item)

print(arr5)
```

## Updating Element of an Array

```python
arr7 = [89, 56, 90, 31, 15, 78]
arr8 = []

for i in range(len(arr7)):
    arr7[3] = 91
    arr8.append(arr7[i])

print(arr8)
```

```python
arr9 = []
for num in arr7:
    arr7[4] = 17
    arr9.append(num)

print(arr9)
```

```python
# Efficient Solution

arr7 = [89, 56, 90, 31, 15, 78]
arr7[3] = 91
arr10 = []
for num in arr7:
    arr10.append(num)

print(arr10)
```

```python
arr7 = [89, 56, 90, 31, 15, 78]

arr7[4] = 17

print(arr7)
```
