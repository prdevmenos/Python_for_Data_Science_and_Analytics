```python
# Sum of n natural numbers.

def findSum(n):
    if (n==0):
        return 0

    res = n*(n+1)
    result = res/2
    return int(result)
```

```python
num = int(input("Enter number"))
findSum(num)
```

```python
num1 = findSum(4)
print(num1)
```

```python
class Solution:
    def FindSum(self, n):
        if n==0:
            return 0

        res = n*(n+1)
        result = res/2
        return int(result)
```

```python
num1 = Solution()

n1 = int(input("Enter a number"))

num1.FindSum(n1)
```

```python
num2= Solution()
num2.FindSum(15)
```
