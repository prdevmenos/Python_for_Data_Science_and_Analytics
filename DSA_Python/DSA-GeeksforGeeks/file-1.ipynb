```python
# Sum of Given Array
```

```python
class Solution:
	def arraySum(self, arr):
	    self.arr = arr
	    sum = 0
	    for i in range(0, len(arr)):
	        sum = sum + arr[i]
	    return sum

Raj = Solution()

arr = list(map(int, input("Enter numbers: ").split()))
result = Raj.arraySum(arr)

print(result)
```

```python
# Check Odd or Even
# Given a positive integer n, determine whether it is odd or even. Return true if the number is 
# even and false if the number is odd.
```

```python
class Solution:
    def isEven (self, n):
        self.n = n
        if n % 2 == 0:
            return "true"
        else:
            return "false"

Prakash = Solution()
n = int(input("Enter a number : "))
result = Prakash.isEven(n)
print(result)
```

```python
# Finding Largest of Given Array.
```

```python
class Solution:
    def largest(self, arr):
        # code here
        self.arr = arr
        l = 0
        for i in range(0, len(arr)):
            if l < arr[i]:
               l = arr[i]
            
        return l

Raj = Solution()

arr = list(map(int, input("Enter numbers: ").split()))
result = Raj.largest(arr)

print(result)
```

```python
# Find numbers from 1 to n without loops. Use (Recursive Functions)
```

```python
class Solution:
    def printTillN(self, n):
    	self.n = n
    	
    	if n == 0:
    	    return
    	
    	self.printTillN(n-1)
    	print(n, end = " ")

Raj = Solution()
n = int(input('Enter Number : '))
Raj.printTillN(n)
```

```python
# Find numbers from n to 1 without Loops. Use (Recursive Functions)
```

```python
class Solve:
    def printRevN(self, n):
        self.n = n

        if n == 0:
            return
        self.printRevN(n - 1)
        print(n, end = " ")

Ram = Solve()
n = int(input("Enter a number: "))
Ram.printRevN(n)
```

```python
class Solution:
    def morefrequent(self, arr, x, y):
        count_x = 0
        count_y = 0

        for i in arr:
            if i == x:
                count_x += 1
        
            elif i == y:
                count_y += 1
        
        if count_x == count_y:
            if x > y:
                return y
            else:
                return x
        elif count_x > count_y:
            return x
        else:
            return y

Ram = Solution()
arr = list(map(int, input("Enter Numbers :").split()))
x = int(input("Enter Number :"))
y = int(input("Enter Number :"))
result = Ram.morefrequent(arr,x,y)
print(result)
```

```python
# To check whether the given input is power of 2 or not
```

```python
class Solution:
    def isPowerofTwo(self, n):
        self.n = n
        if (n <= 0):
            return False
        
        num = 0
        while 2 ** num <= n:
            if n == 2 ** num:
                return True
            num += 1
        else:
            return False
            
Pow = Solution()
n = int(input("Enter Number :"))
result = Pow.isPowerofTwo(n)
print(result)
```

```python
# Count odd and even from an array.
```

```python
class Solution:
    def countOddEven(self, arr):
        count_odd = 0
        count_even = 0
        for num in arr:
            if num%2 == 0:
                count_even += 1
            else:
                count_odd += 1
        return count_odd, count_even

C = Solution()
arr = list(map(int, input().split()))
result = C.countOddEven(arr)
print(result)
		
```
