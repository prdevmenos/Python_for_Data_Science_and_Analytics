```python
# Two Sum for Sorted Array.
```

```python
class Solution(object):
    def twoSum(self, numbers, target):
        self.numbers = numbers
        self.target = target
        i = 0
        j = len(numbers) - 1

        while(i > j): # this is mistake because it is sorted array i pointer less than j pointer. 
            sum = numbers[i] + numbers[j]
            if(sum == target):
                return[i,j]
            
            if(sum > target):
                j -= 1
            else:
                i += 1

Raj = Solution()

numbers = list(map(int, input().split()))
target = int(input("Enter target: "))
result = Raj.twoSum(numbers, target)
print(result)     
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        self.nums = nums
        self.target = target
        i = 0
        j = len(nums) - 1
        while(i < j):
            sum = nums[i] + nums[j]
            if(sum == target):
                return(i,j)
            if(sum < target):
                i += 1
            else:
                j -= 1

Raj = Solution()

nums = list(map(int, input().split()))
target = int(input("Enter target: "))
result = Raj.twoSum(nums, target)
print(result)
```

```python
# Two Sum for Unsorted Array
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        seen = {}

        for i in range(len(nums)):
            complement = target - nums[i]

            if complement in seen:
                return [seen[complement], i]

            seen[nums[i]] = i

Raj = Solution()

nums = list(map(int, input().split()))
target = int(input("Enter target: "))
result = Raj.twoSum(nums, target)
print(result)
```

```python
# Two Sum Sorted Array Using For Loop.
```

```python
class Solve(object):
    def Prasanna(self, numbers, target):
        self.numbers = numbers
        self.target = target

        for i in range(len(numbers)):
            for j in range(i + 1, len(numbers)):
                total = numbers[i] + numbers[j]

                if total == target:
                    return (i, j)

        return None


Raj = Solve()

numbers = list(map(int, input("Enter numbers: ").split()))
target = int(input("Enter target: "))

result = Raj.Prasanna(numbers, target)

print(result)
```
