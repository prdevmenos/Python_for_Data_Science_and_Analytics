### Conditional Statement

```python
n = 10
if n == 10:
    print("ten")
else:
    print("not ten")
```

```python
n = int(input('Enter number :'))
if n % 2 == 0:
    print(n,'is even')
else:
    print(n,'is odd')
```

### Loops

```python
for i in range(1,5,1):
    print(i)
```

```python
names = ['pratyush', 'raj', 'meera']
for name in names:
    print(name)
```

```python
sum = 0
for i in range(1,6):
    sum = sum + i

print("Sum of Numbers are :",sum)
```

```python
# Sum of 10 numbers are:
sum1 = 0
for j in range(1,11):
    sum1 = sum1 + j
print("Sum of 10 numbers are", sum1)
```

```python
# Sum of first 10 even numbers
sum = 0
for i in range(1,21):
    if i%2 == 0:
        sum = sum + i
        print(i)
print("The sum of first 10 even number is",sum)
```

```python
# Sum of even numbers between 20 and 40(both included).

total_sum = 0
for i in range(20, 41):
    if i%2 == 0:
        total_sum = total_sum + i
        print(i)

print(f"Sum of even numbers between 20 and 40 is {total_sum}")
```
